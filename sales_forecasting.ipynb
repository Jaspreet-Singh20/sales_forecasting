```python
sales = pd.read_csv("sales.csv")
stores = pd.read_csv("stores.csv")
calendar = pd.read_csv("calendar.csv")
customers = pd.read_csv("customers.csv")
product = pd.read_csv("products.csv")
```

```python
print(sales.head())
print(stores.head())
print(calendar.head())
print(customers.head())
print(product.head())
```

```python
print(sales.columns)
print(stores.columns)
print(calendar.columns)
print(customers.columns)
print(product.columns)
```

```python
sales['order_date'] = pd.to_datetime(
    sales['order_date']
)

calendar['date'] = pd.to_datetime(
    calendar['date']
)

customers['join_date'] = pd.to_datetime(
    customers['join_date']
)
```

```python
df = sales.merge(
    stores,
    on='store_id',
    how='left'
)
```

```python
df = df.merge(
    customers,
    on='customer_id',
    how='left'
)
```

```python
df = df.merge(
    product,
    on='product_id',
    how='left'
)
```

```python
df = df.merge(
    calendar,
    left_on='order_date',
    right_on='date',
    how='left'
)
```
